# Chat Word Treatment

```python
chat_words = {
    'u': 'you',
    'r': 'are',
    'ur': 'your',
    'btw': 'by the way',
    'lol': 'laughing out loud',
    'omg': 'oh my god'
}
```

```python
def chat_conversion(text):
    new_text=[]
    for word in text.split():
        if word in chat_words:
            new_text.append(chat_words[word])
        else:
            new_text.append(word)
    return " ".join(new_text)
```

```python
print(chat_conversion("omg u r awesome"))
```

```python
from nltk.stem import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
print(ps.stem("running"))
print(ps.stem("studies"))
print(ps.stem("playing"))
```

```python
tokens = ['running', 'studies', 'playing']

stemmed=[ps.stem(word) for word in tokens]
print(stemmed)
```

```python
def stem_words(tokens):
    return [ps.stem(word) for word in tokens]
```

```python
data['review']=data['review'].apply(stem_words)
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lammatizer=WordNetLemmatizer()
```

```python
print(ps.stem('studies'))
print(lammatizer.lemmatize('studies'))

print(ps.stem('running'))
print(lammatizer.lemmatize('running'))
print(lammatizer.lemmatize('running', pos='v'))
```

```python
def lemmatizer_words(tokens):
    return [lammatizer.lemmatize(word) for word in tokens]
```

```python
data['review_lemmatized'] = data['review'].apply(lemmatizer_words)
```
