```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
from string import punctuation
```

```python
stop=list(STOP_WORDS)
```

```python
len(stop)
```

```python
stop
```

```python
doc=nlp(text)
```

```python
doc
```

```python
tokens=[token.text for token in doc]
```

```python
print(tokens)
```

```python
doc
```

```python
wordfreq={}

for word in doc:
    if word.text.lower() not in stop:
        if word.text.lower() not in punctuation:
            if word.text not in wordfreq.keys():
                wordfreq[word.text]=1
            else:
                wordfreq[word.text]+=1
```

```python
wordfreq
```

```python
len(wordfreq)
```

```python
maxfreq=max(wordfreq.values())
maxfreq
```

```python
for word in wordfreq.keys():
    wordfreq[word]=wordfreq[word]/maxfreq
```

```python
wordfreq
```

```python
maxfreq=max(wordfreq.values())
maxfreq
```

```python
sentt=[sent for sent in doc.sents]
sentt
```

```python
len(sentt)
```

```python
# indivisualscore/maxscore
```

```python
sentscore={}

for sent in sentt:
    for word in sent:
        if word.text.lower() in wordfreq.keys():
            if sent not in sentscore.keys():
                sentscore[sent]=wordfreq[word.text.lower()]
            else:
                sentscore[sent]+=wordfreq[word.text.lower()]
```

```python
print(sentscore)
```

```python
len(sentscore)
```

```python
print(max(sentscore.values()))
```
