```python
from nltk.util import bigrams,trigrams,ngrams
import nltk
```

```python
string="""hello the best and most beautifull thing i
n the world cannot be seen or even touched, they must be felt with heart  
"""
```

```python
from nltk.tokenize import word_tokenize
```

```python
tokens=word_tokenize(string)
```

```python
print(tokens)
```

```python
bigram=list(bigrams(tokens))
```

```python
bigram
```

```python
trigram=list(trigrams(tokens))
```

```python
trigram
```

```python
ngram=list(ngrams(tokens,5))
```

```python
ngram
```

# stem

```python
from nltk.stem import PorterStemmer
pst=PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximuim')
```

```python
words=['give','giving','given','gave']
for word in words:
    print(word , " : ",pst.stem(word))
```

```python
from nltk.stem import LancasterStemmer
lst=LancasterStemmer()
```

```python
for word in words:
    print(word , " : ",lst.stem(word))
```

```python
from nltk.stem import SnowballStemmer
```

```python
sbst=SnowballStemmer('english')
```

```python
for word in words:
    print(word , " : ",sbst.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lem=WordNetLemmatizer()
```

```python
for word in words:
    print(word , " : ",lem.lemmatize(word))
```

# pos tags

```python
sent = 'kathy is a natural when it comes to drawing'
```

```python
from nltk.tokenize import sent_tokenize,word_tokenize
```

```python
tokens=word_tokenize(sent)
```

```python
tokens
```

```python
for token in tokens:
    print(nltk.pos_tag([token]))
```

```python
sent2 = 'john is eating a delicious cake'
tokens2=word_tokenize(sent2)
for token in tokens2:
    print(nltk.pos_tag([token]))
```

# NER (Named Entity Recogonition)

```python
from nltk import ne_chunk
```

```python
sent = 'The US president stays in the WHITEHOUSE '
```

```python
tokens3=word_tokenize(sent)
```

```python
tokentag=nltk.pos_tag(tokens3)
```

```python
tokentag
```

```python
tokenner=ne_chunk(tokentag)
```

```python
print(tokenner)
```
